```python
# LOS distribution investigation
print("="*60)
print("Length of Stay (LOS) Distribution")
print("-"*60)

# Counts based on thresholds
stays_lt_1_day = (df["los"] < 1).sum()
stays_gt_7_days = (df["los"] > 7).sum()
stays_gt_14_days = (df["los"] > 14).sum()

print(f" ICU stays < 1 day:   {stays_lt_1_day:,}")
print(f" ICU stays > 7 days:  {stays_gt_7_days:,}")
print(f" ICU stays > 14 days: {stays_gt_14_days:,}")

# Shortest stay + quick clinical check
shortest_los = df["los"].min()
shortest_hours = shortest_los * 24

print("-"*60)
print(f" Shortest ICU stay: {shortest_los:.6f} days (~{shortest_hours:.2f} hours)")

# Clinical sense check (brief interpretation)
if shortest_los < 0.1:
    print(" Clinical note: Very short ICU stays can happen (rapid transfer/discharge),")
    print("   but they may also reflect brief administrative stays or unit transfers.")
else:
    print(" Clinical note: This duration can be clinically plausible for short observation stays.")
print("="*60)
```

### 3.3 Transfer Validation (5 points)

```python
# Investigating ICU Unit Transfers
print("="*60)
print(" ICU Unit Transfer Analysis")
print("-"*60)

# Identify stays where patient transferred between ICU units
transfers = df[df["first_careunit"] != df["last_careunit"]]

# Count number of transfer stays
num_transfers = len(transfers)

print(f" ICU stays involving a unit transfer: {num_transfers:,}")

print("-"*60)

# Identify most common transfer patterns
transfer_patterns = (
    transfers.groupby(["first_careunit", "last_careunit"])
    .size()
    .sort_values(ascending=False)
)

print(" Most Common Transfer Patterns:")
print(transfer_patterns)

print("="*60)
```

**Your Findings:**

- What percentage of stays involve an ICU transfer?
Total ICU stays: 140
ICU stays involving transfer: 11
11/140 x 100= 7.9
  This suggests that most patients remain in the same ICU type throughout their stay, but a small portion require movement between specialty units.
- Why might a patient be transferred from one ICU type to another?
Patients may be transferred between ICU units for several clinical reasons:
 .Change in clinical condition (e.g., cardiac complications requiring CCU monitoring)
 .Escalation of care to a more specialized unit
 .De-escalation of care to a lower-acuity unit (e.g., Neuro ICU → Neuro Stepdown)
 .Post-surgical transition (e.g., Surgical ICU → Medical ICU for ongoing medical management)
 .Development of new complications requiring specialty expertise
  Clinically, ICU transfers often reflect changes in patient acuity or evolving medical needs. They may also indicate complex or unstable cases requiring multiple specialty teams.

## Part 4: Create ICU Type Groups (10 points)

```python
# Inspect unique ICU unit names in first_careunit
print("="*60)
print("🔎 Unique values in first_careunit")
print("-"*60)

unique_units = df["first_careunit"].unique()
print(unique_units)

print("="*60)
```

```python
#  Create icu_category column using Version B specialty categories
print("="*60)
print(" Creating ICU Specialty Categories (Version B)")
print("-"*60)

icu_mapping = {
    "Medical Intensive Care Unit (MICU)": "Medical",
    "Medical/Surgical Intensive Care Unit (MICU/SICU)": "Medical",
    
    "Surgical Intensive Care Unit (SICU)": "Surgical",
    "Trauma SICU (TSICU)": "Surgical",
    
    "Coronary Care Unit (CCU)": "Cardiac",
    "Cardiac Vascular Intensive Care Unit (CVICU)": "Cardiac",
    
    "Neuro Surgical Intensive Care Unit (Neuro SICU)": "Neurological",
    "Neuro Stepdown": "Neurological",
    "Neuro Intermediate": "Neurological"
}

# Map first_careunit to broader clinical category
df["icu_category"] = df["first_careunit"].map(icu_mapping)

# Check for any units that did not map correctly
unmapped = df[df["icu_category"].isna()]["first_careunit"].unique()

print(" icu_category column created.")
print(f" Unmapped ICU units (if any): {unmapped}")

print("="*60)
```

```python
# Verify groupings worked (counts per ICU category)
print("="*60)
print(" ICU Category Counts")
print("-"*60)

category_counts = df["icu_category"].value_counts(dropna=False)
print(category_counts)

print("="*60)
```

```python
# Calculate average LOS for each ICU specialty category
print("="*60)
print(" Average LOS by ICU Specialty Category")
print("-"*60)

avg_los_by_category = df.groupby("icu_category")["los"].mean().sort_values(ascending=False)
print(avg_los_by_category)

print("="*60)
```
